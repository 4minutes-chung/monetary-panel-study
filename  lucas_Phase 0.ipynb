```python
# 📁 Step 1: Load Your Files (edit path if needed)

# Update the paths to where your downloaded files are
m2_raw = pd.read_csv("m2_raw.csv", skiprows=4)
cpi_raw = pd.read_csv("cpi_raw.csv", skiprows=4)
gdp_raw = pd.read_csv("gdp_raw.csv", skiprows=4)
```

```python
# 🧼 Step 3: Tidy Function to Clean All Datasets

def clean_world_bank(df_raw, value_name):
    """Convert wide WB CSV into tidy format: country, year, value"""
    df = df_raw.drop(columns=["Indicator Name", "Indicator Code"], errors='ignore')
    df = df.melt(id_vars=["Country Name", "Country Code"], var_name="year", value_name=value_name)
    
    df = df[df["year"].str.isnumeric()]
    df["year"] = df["year"].astype(int)
    df[value_name] = pd.to_numeric(df[value_name], errors="coerce")
    df = df[df[value_name].notna()]  
    return df

# Apply to all 3 datasets
df_m2 = clean_world_bank(m2_raw, "m2")
df_cpi = clean_world_bank(cpi_raw, "cpi")
df_gdp = clean_world_bank(gdp_raw, "gdp")
```

```python
def compute_log_growth(df, value_col, new_colname):
    """
    Compute log-difference growth rates per country
    """
    df = df.copy()
    df.sort_values(by=["Country Name", "year"], inplace=True)
    df[new_colname] = df.groupby("Country Name")[value_col].transform(lambda x: np.log(x) - np.log(x.shift(1)))
    return df.dropna(subset=[new_colname])
```

```python
df_m2_growth  = compute_log_growth(df_m2, "m2", "m2_growth")
df_cpi_growth = compute_log_growth(df_cpi, "cpi", "inflation")
df_gdp_growth = compute_log_growth(df_gdp, "gdp", "gdp_growth")
```

## Lucas-Style 30-Year Averages (Long-Run Cross-Country)
For each country, compute average M2 growth, average inflation, average real GDP growth over a consistent ~30-year period.
Problem: Not all countries have full 1960–1990 or 1994–2024 data
Fix: Dynamic 30-Year Window Per Country
	1.	Only include countries with at least 25+ years of valid data
	2.	Use each country’s longest 30-year span, or close to it
	3.	Compute means only within that window

```python
def compute_country_averages(df, varlist, min_years=25):
    """
    Computes per-country long-run averages over available data,
    filtering out countries with too few observations.
    """
    results = []

    for country, group in df.groupby("Country Name"):
        if group.shape[0] >= min_years:
            avg_vals = group[varlist].mean()
            avg_vals["Country Name"] = country
            avg_vals["n_obs"] = group.shape[0]
            results.append(avg_vals)

    df_avg = pd.DataFrame(results)
    return df_avg.reset_index(drop=True)
```

```python
vars_to_avg = ["m2_growth", "inflation", "gdp_growth"]
df_lucas = compute_country_averages(df_merged, vars_to_avg)
df_lucas.head()
```

```python
sns.regplot(data=df_lucas, x="m2_growth", y="inflation")
plt.title("Average M2 Growth vs Average Inflation")
plt.show()

sns.regplot(data=df_lucas, x="m2_growth", y="gdp_growth")
plt.title("Average M2 Growth vs Average GDP Growth")
plt.show()
```

```python
import statsmodels.api as sm

X = sm.add_constant(df_lucas["m2_growth"])
y1 = df_lucas["inflation"]
y2 = df_lucas["gdp_growth"]

model1 = sm.OLS(y1, X).fit()
print(model1.summary())

model2 = sm.OLS(y2, X).fit()
print(model2.summary())
```

✅ INTERPRETATION:
	•	Strong coefficient + R² on M2 → Inflation = Lucas’s point: “Money growth drives inflation cross-country in the long run.”
	•	No correlation with GDP growth = Monetary neutrality long-run.

inflation = -0.044 + 0.905 × m2_growth
R² = 0.885, p < 0.001

🔍 What it means:
	•	✅ Strong, positive, significant relationship.
	•	✅ R² of 0.885 means 88.5% of cross-country variation in inflation is explained by average money growth over the ~30-year sample window.
	•	✅ The slope ≈ 0.9 says:
A 1% increase in average M2 growth is associated with a 0.9% increase in average inflation.
	•	✅ p-value = 0.000 → this is statistically rock-solid.

🔗 Matches Lucas (1996) exactly. ✔️

gdp_growth = 0.0374 - 0.0016 × m2_growth
R² ≈ 0.000, p = 0.902

🔍 What it means:
	•	❌ No relationship at all.
	•	❌ R² = 0.000 → money growth explains nothing about GDP growth across countries.
	•	❌ Slope ≈ 0 (−0.0016) and p = 0.902 → not statistically significant.

🔗 This is the Lucas punchline:

Money growth drives inflation, not real GDP.
Long-run neutrality of money holds across countries.
